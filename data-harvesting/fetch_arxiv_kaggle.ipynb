```python
from constant import data_path
path =  f"{data_path}/arxiv/arxiv-metadata-oai-snapshot.json"
import polars as pl
df = pl.scan_ndjson(path)
df.head().collect(engine="streaming")
```

```python
df.with_columns(
    pl.col("doi").is_not_null().sum(),
    pl.col("id").unique().count()
).collect(engine="streaming")
```

```python
df.head(1).collect(engine="streaming").get_column("versions").to_list()[0]
```

```python
df.with_columns(
    pl.col("categories").str.split(" ").alias("categories"),
    ("10.48550/arxiv." + pl.col("id")).alias("doi"),
    pl.col("doi").alias("curated_doi"),

    pl.col("versions").list.last().struct.field("created")
    .str.strptime(pl.Datetime, "%a, %d %b %Y %H:%M:%S %Z", strict=False)
    .dt.strftime("%Y-%m-%d")
    .alias("last_updated_date"),

    pl.col("versions").list.first().struct.field("created")
    .str.strptime(pl.Datetime, "%a, %d %b %Y %H:%M:%S %Z", strict=False)
    .dt.strftime("%Y-%m-%d")
    .alias("published_date")
).with_columns(
    pl.col("categories").list.first().alias("primary_category")
).select([
    pl.col("id"),
    pl.col("doi"),
    pl.col("categories"),
    pl.col("primary_category"),
    pl.col("curated_doi"),
    pl.col("published_date"),
    pl.col("last_updated_date"),
]).sink_ndjson(f"{data_path}/arxiv/arxiv_preprints.dedup.jsonl")
```
